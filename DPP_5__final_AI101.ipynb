### ▶ **Creating NumPy Arrays**
NumPy is the core library for numerical computing in Python. Its main object is the n-dimensional array, which is much faster and more memory-efficient than a normal Python list when doing numerical operations.

**Why not just use lists?**

Python lists can hold mixed data types and are flexible, but that flexibility makes them slow for math operations. NumPy arrays store data of a single type in contiguous memory, so operations on them are vectorized and fast.

First, we need to import the library. The common convention is to import it as `np`.

````python
import numpy as np
# Creating a 1D array from a list
a = np.array([10, 20, 30, 40, 50])
print(a)
print(type(a))
````

**Creating 2D arrays (matrices)**

You can pass a list of lists to `np.array()` to create a 2D array.

````python
b = np.array([[1, 2, 3], [4, 5, 6]])
print(b)
````

**Other handy ways to create arrays**

 `np.zeros(shape)` – array filled with 0's

 `np.ones(shape)` – array filled with 1's

 `np.arange(start, stop, step)` – like Python's `range()` but returns an array

 `np.linspace(start, stop, num)` – `num` evenly spaced values between start and stop

 `np.random.rand(shape)` – array of random values between 0 and 1

````python
zeros_arr = np.zeros((2, 3))
ones_arr = np.ones((3, 3))
range_arr = np.arange(0, 10, 2)
lin_arr = np.linspace(0, 1, 5)
print("Zeros:\n", zeros_arr)
print("Ones:\n", ones_arr)
print("Arange:", range_arr)
print("Linspace:", lin_arr)
````

**Array attributes**Every NumPy array has some useful attributes that tell you about its structure:* `.shape` – dimensions of the array (rows, columns)* `.ndim` – number of dimensions* `.size` – total number of elements* `.dtype` – data type of the elements

````python
print("Shape:", b.shape)
print("Dimensions:", b.ndim)
print("Size:", b.size)
print("Dtype:", b.dtype)
````

**Reshaping arrays**`.reshape()` lets you change the shape of an array without changing its data, as long as the total number of elements stays the same.

````python
c = np.arange(1, 13)
print("Original:", c)
c_reshaped = c.reshape(3, 4)
print("Reshaped (3x4):\n", c_reshaped)
````

**Notice something?** Columns like `discounted_price`, `actual_price`, `discount_percentage`, `rating`, and `rating_count` all show up as `object` (text) dtype instead of numbers! That's because they contain symbols like `₹`, `%`, and `,` — very common in real, uncleaned CSV data. We can't compute a mean or filter numerically on a text column, so let's clean these up next.

### ▶ **A Quick Data Cleaning Step**

We'll strip out the extra symbols and convert these columns to proper numeric types using `.str.replace()` and `pd.to_numeric()`. Setting `errors="coerce"` turns any value that still can't be converted (like a stray `|` in the ratings column) into `NaN` instead of crashing.

````python
#This line cleans up the discounted_price column, which stores values like "₹1,099" as plain text. The first .str.replace("₹", "", regex=False) strips out the rupee symbol, turning "₹399" into "399".
#That result is immediately chained into a second .str.replace(",", "", regex=False), which removes commas, turning "1,099" into "1099"
#Once the string is a clean number, .astype(float) converts it from text into an actual numeric type so it can be used in math, comparisons, and aggregation.
#The regex=False argument just tells pandas to treat "₹" and "," as literal characters rather than regex patterns.
df["discounted_price"] = df["discounted_price"].str.replace("₹", "", regex=False).str.replace(",", "", regex=False).astype(float)
df["actual_price"] = df["actual_price"].str.replace("₹", "", regex=False).str.replace(",", "", regex=False).astype(float)
df["discount_percentage"] = df["discount_percentage"].str.replace("%", "", regex=False).astype(float)
df["rating_count"] = df["rating_count"].str.replace(",", "", regex=False)
#This converts the now comma-free rating_count strings into numbers using pd.to_numeric() instead of .astype(float).
#The reason for the switch is that rating_count has a couple of missing or blank entries, which would cause .astype(float) to crash.
#pd.to_numeric() with errors="coerce" is more forgiving — anything it can't convert becomes NaN instead of raising an error, so the rest of the column still converts successfully.
df["rating_count"] = pd.to_numeric(df["rating_count"], errors="coerce")
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
````

Much better — these are now `float64` columns we can actually do math on. The `category` column is also a bit messy: it stores a full hierarchy like `Electronics|Mobiles|SmartWatches` separated by `|`. Let's pull out just the top-level category for easier grouping later.

````python
# This line creates a new column called "main_category" from the existing "category" column.
# The .str.split("|") splits each category wherever the "|" symbol appears.
# For example, "Electronics|Mobiles|Smartphones" becomes ["Electronics", "Mobiles", "Smartphones"].
# The .str[0] selects the first element, "Electronics", which represents the main category.
# This value is then stored in the new "main_category" column.

df["main_category"] = df["category"].str.split("|").str[0]

# This line counts how many times each unique main category appears in the "main_category" column.
# It helps us understand how many products belong to each main category.
# The results are then printed on the screen.

print(df["main_category"].value_counts())
````

### ▶ **Pandas Series**
A **Series** is a one-dimensional labeled array — think of it as a single column of a DataFrame, or a supercharged list where each value has an associated index label.

````python
marks = pd.Series([85, 90, 78, 92], index=["Math", "Physics", "Chemistry", "Biology"])
print(marks)
````

You can access elements by their **label** or by their **position**.

````python
print(marks["Physics"])   # by label
print(marks.iloc[1])       # by position
````

A single column of a DataFrame is itself a Series.

````python
ratings = df["rating"]
print(ratings)
print(type(ratings))
````

### ▶ **Pandas DataFrames**

### ▶ **Basic Indexing**

Pandas gives you a few different ways to select data from the Amazon DataFrame (`df`):
* `df["col"]` or `df[["col1", "col2"]]` – select column(s)
* `df.loc[]` – select rows/columns by **label**
* `df.iloc[]` – select rows/columns by **integer position**

````python
# Selecting a single column
print(df["product_name"])
````

````python
# Selecting multiple columns
print(df[["product_name", "discounted_price", "rating"]])
````

````python
# loc - label based
print(df.loc[0])                                        # first row (by label/index)
print(df.loc[0, "product_name"])                        # value at row 0, column 'product_name'
print(df.loc[0:2, ["product_name", "rating"]])          # slice of rows, specific columns
````

````python
# iloc - position based
print(df.iloc[0])          # first row
print(df.iloc[0, 1])       # row 0, column at position 1 (product_name)
print(df.iloc[0:3, 0:3])   # first 3 rows, first 3 columns
````

### ▶ **Filtering**

You can filter rows using **boolean conditions**, similar to how you'd filter a list with a condition, but pandas does it in a vectorized way across the whole column.

````python
# Products rated above 4.5
print(df[df["rating"] > 4.5])
````

````python
# Products in the Electronics main category
print(df[df["main_category"] == "Electronics"])
````

You can also combine multiple conditions using `&` (and), `|` (or). Each condition must be wrapped in parentheses.

````python
# Electronics products with a discount of over 50%
print(df[(df["main_category"] == "Electronics") & (df["discount_percentage"] > 50)])
````

### ▶ **Summary Operations**

Pandas provides quick built-in methods to summarize your data:
* `.describe()` – summary statistics for numeric columns
* `.mean()`, `.sum()`, `.min()`, `.max()`, `.median()`, `.std()` – column-wise stats
* `.value_counts()` – frequency count of unique values (useful on a single column)

````python
print(df.describe())
````

````python
print("Average Rating:", df["rating"].mean())
print("Highest Discounted Price:", df["discounted_price"].max())
print("Lowest Discounted Price:", df["discounted_price"].min())
print("Average Discount %:", df["discount_percentage"].mean())
````

````python
print(df["main_category"].value_counts())
````
